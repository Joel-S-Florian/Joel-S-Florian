```python
from textblob import TextBlob
import time

# 1. DATASET: Lista de comentarios en español
comentarios = [
    "El talento lo llevas en la sangre, solo falta que te circule", # Tu frase reto
    "Este es el mejor curso de IA que he tomado, ¡increíble!",
    "La comida estaba fría y el servicio fue pésimo.",
    "Mañana tengo una reunión a las 5 de la tarde.",
    "No estoy seguro de si me gusta, es un poco extraño."
]

def clasificador_pro(texto):
    try:
        # Creamos el objeto TextBlob con tu frase
        blob = TextBlob(texto)

        # SUPERPODER: Traducir al inglés para máxima precisión de la IA
        # TextBlob usa la API de Google Translate por detrás
        blob_en = blob.translate(from_lang='es', to='en')

        # Obtenemos la polaridad (-1 a 1)
        polaridad = blob_en.sentiment.polarity

        # 2. LÓGICA DE CLASIFICACIÓN (Tu condicional mejorada)
        if polaridad > 0.2:
            return "POSITIVO 😊", polaridad
        elif polaridad < -0.2:
            return "NEGATIVO 😡", polaridad
        else:
            return "NEUTRO 😐", polaridad

    except Exception:
        # Si algo falla (como el internet), devolvemos un estado seguro
        return "ERROR / NEUTRO ⚪", 0

# 3. EJECUCIÓN Y PRESENTACIÓN
print("="*40)
print("   IA SENTIMENT ANALYSIS PRO v1.0   ")
print("="*40)

for frase in comentarios:
    resultado, puntaje = clasificador_pro(frase)
    print(f"📝 Frase: '{frase}'")
    print(f"📊 Resultado: {resultado} (Puntaje: {puntaje:.2f})")
    print("-" * 40)
    time.sleep(0.5) # Pausa dramática para que parezca que la IA "piensa"
```
